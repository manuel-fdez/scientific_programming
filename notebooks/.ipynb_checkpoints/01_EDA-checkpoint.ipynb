Vamos a comprobar si el dataset está balanceado. Para ello, vamos a ver la distribución de la variable objetivo.

```python
renamed_df["diagnosis_label"] = renamed_df["diagnosis"].map(lambda x: "benign" if x=="1" else "malignant") # Using quotion marks because diagnosis is categorical
sns.countplot(data=renamed_df, x="diagnosis_label")
plt.xlabel("Diagnosis")
plt.ylabel("Total")
plt.show()
```

```python
#Print the percentages of each class
value_counts = renamed_df["diagnosis_label"].value_counts()

plt.pie(value_counts.values, labels=value_counts.index, autopct='%1.1f%%')
plt.show()
```

Vemos que el dataset no está balanceado, ya que hay más casos benignos que malignos.

Procedemos a chequear si hay outliers en las variables numéricas usando boxplots.

```python
# We get only the numerical columns
numerical_df = renamed_df.select_dtypes(include=[np.number])
```

Tenemos 30 variables numéricas en total. Vamos a echar un vistazo imprimiendo todos los boxplots juntos.

```python
plt.figure(figsize=(18, 6))
numerical_df.boxplot(rot=90)
plt.title("Boxplots of numerical features")
plt.tight_layout()
plt.show()
```

Al tener escalas diferentes, es difícil ver los outliers en algunos casos. Podemos ver que hay outliers en muchas de las variables. Para facilitar su visualización, vamos a escalar las variables numéricas usando StandardScaler de sklearn, que centra las variables en la media y las escala a desviación estándar 1.

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
numerical_scaled = scaler.fit_transform(numerical_df)

numerical_scaled_df = pd.DataFrame(
    numerical_scaled,
    columns=numerical_df.columns,
    index=numerical_df.index
)
```

```python
plt.figure(figsize=(18, 6))
numerical_scaled_df.boxplot(rot=90)
plt.title("Boxplots of numerical features (scaled)")
plt.tight_layout()
plt.show()
```

Vamos a ver las correlaciones entre las variables numéricas, por si es necesario eliminar alguna variable muy correlacionada con otra.

```python
# Calculate correlation matrix
corr_matrix = numerical_df.corr()
corr_matrix
```

```python
# Como la tabla es muy grande, vamos a visualizarla con un heatmap de seaborn.
plt.figure(figsize=(15, 12))
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', center=0
              , cbar_kws={"shrink": .8})                
plt.title("Correlation Matrix Heatmap")
plt.show()
```

Vemos que hay variables altamente correlacionadas. Vamos a considerar como altamente correlacionadas aquellas con un valor absoluto de correlación mayor a 0.9.

```python
threshold = 0.9

high_corr = (
    corr_matrix.abs()
    .where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    .stack()
    .reset_index()
)
high_corr.columns = ["feature_1", "feature_2", "correlation"]

high_corr[high_corr["correlation"] > threshold].sort_values("feature_1", ascending=True).reset_index(drop=True)
```

The correlation analysis reveals strong redundancy among several groups of features, particularly those related to tumor size (radius, perimeter, area), concavity measures, and their associated error and worst-case variants.
